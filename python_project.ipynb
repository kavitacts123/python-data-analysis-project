```python
# Category-wise Sales (Bar Chart)
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["TotalAmount"].sum()

plt.figure(figsize=(4,3))

# Attractive colors
colors = ["#4E79A7", "#F28E2B", "#59A14F", "#E15759", "#76B7B2"]

bars = category_sales.plot(
    kind="bar",
    color=colors,
    edgecolor="black",
    linewidth=1.2
)

# Value labels on top of bars
for i, value in enumerate(category_sales):
    plt.text(i, value + 50, f"{value:,.0f}",
             ha="center", fontsize=6, fontweight="bold")

plt.title("Category-wise Sales Analysis",
          fontsize=12,
          fontweight="bold",
          color="darkblue")

plt.xlabel("Product Category",
           fontsize=10,
           fontweight="bold")

plt.ylabel("Total Sales (₹)",
           fontsize=10,
           fontweight="bold")

plt.xticks(rotation=20, fontsize=8)
plt.yticks(fontsize=10)

# Grid
# plt.grid(axis="y", linestyle="--", alpha=0.5)

# Remove top & right border
plt.box(False)

plt.tight_layout()

plt.savefig("category_wise.png", dpi=100, bbox_inches="tight")

plt.show()  
```

```python
import matplotlib.pyplot as plt

# Top 10 Products (Bar Chart)
top_products = (
    df.groupby("Product")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
    .head(6)
)

# Figure Size
plt.figure(figsize=(6, 4))

# Attractive Colors (10 colors)
colors = [
    "#4E79A7", "#F28E2B", "#59A14F", "#E15759", "#76B7B2",
    "#EDC948"
]

# Plot
bars = top_products.plot(
    kind="bar",
    color=colors,
    edgecolor="black",
    linewidth=1.2,
    width=0.7
)

# Value Labels
for i, value in enumerate(top_products):
    plt.text(
        i,
        value,
        f"{value:,.0f}",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold"
    )

# Title & Labels
plt.title(
    "Top 10 Products by Sales",
    fontsize=12,
    fontweight="bold",
    color="darkblue"
)

plt.xlabel(
    "Products",
    fontsize=10,
    fontweight="bold"
)

plt.ylabel(
    "Total Sales (₹)",
    fontsize=10,
    fontweight="bold"
)

# X-axis Rotation
plt.xticks(rotation=45, ha="right", fontsize=9)
plt.yticks(fontsize=10)

# Grid
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Remove extra borders
plt.box(False)

plt.tight_layout()

# Save
plt.savefig("top10_products.png", dpi=100, bbox_inches="tight")

plt.show()
```

```python
# Platform-wise Sales (Pie Chart)
import matplotlib.pyplot as plt

# Platform-wise Sales
platform_sales = df.groupby("Platform")["TotalAmount"].sum()

# Figure Size
plt.figure(figsize=(4, 4))

# Attractive Colors
colors = ["#4E79A7", "#F28E2B", "#59A14F", "#E15759", "#76B7B2"]

# Pie Chart
platform_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90,
    colors=colors,
    shadow=True,
    explode=[0.05] * len(platform_sales),   # Slightly separate all slices
    wedgeprops={"edgecolor": "white", "linewidth": 2},
    textprops={"fontsize": 10, "fontweight": "bold"}
)

# Title
plt.title(
    "Platform-wise Sales",
    fontsize=16,
    fontweight="bold",
    color="darkblue"
)

# Remove Y-axis Label
plt.ylabel("")

# Equal Circle
plt.axis("equal")

# Save Chart
plt.tight_layout()
plt.savefig("platform_sales.png", dpi=100, bbox_inches="tight")

plt.show()
```

```python
# City-wise Sales (Bar Chart)
import matplotlib.pyplot as plt

# Top 3 Cities by Sales
city_sales = (
    df.groupby("City")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
    .head(3)
)

plt.figure(figsize=(6, 5))

# Attractive Colors
colors = ["#4E79A7", "#F28E2B", "#59A14F"]

# Bar Chart
bars = city_sales.plot(
    kind="bar",
    color=colors,
    edgecolor="black",
    linewidth=1.2,
    width=0.6
)

# Value Labels
for i, value in enumerate(city_sales):
    plt.text(
        i,
        value,
        f"{value:,.0f}",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

# Title & Labels
plt.title(
    "Top 3 Cities by Sales",
    fontsize=15,
    fontweight="bold",
    color="darkblue"
)

plt.xlabel("City", fontsize=12, fontweight="bold")
plt.ylabel("Total Sales (₹)", fontsize=12, fontweight="bold")

plt.xticks(rotation=0, fontsize=10)
plt.yticks(fontsize=10)

# Grid
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Remove extra borders
plt.box(False)

plt.tight_layout()

plt.savefig("top3_city_sales.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Price Distribution (Histogram)
import matplotlib.pyplot as plt

# Figure Size
plt.figure(figsize=(6, 4))

# Histogram
plt.hist(
    df["Price"],
    bins=10,
    color="#4E79A7",
    edgecolor="black",
    linewidth=1.2,
    alpha=0.8
)

# Title
plt.title(
    "Price Distribution",
    fontsize=16,
    fontweight="bold",
    color="darkblue"
)

# Axis Labels
plt.xlabel(
    "Price (₹)",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Frequency",
    fontsize=12,
    fontweight="bold"
)

# Grid
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Remove top & right borders
plt.box(False)

# Layout
plt.tight_layout()

# Save Chart
plt.savefig("price_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```
